```python
# @title 🚀 Cell 2: Launch KoboldCpp + Stheno 8B Server
# @markdown Pilih versi quant (Q5_K_M sangat direkomendasikan untuk kualitas terbaik di 16GB VRAM):
quant_quality = "Q5_K_M" #@param ["Q5_K_M", "Q4_K_M"]
context_size = 16384 #@param [8192, 16384, 24576, 32768]

import os
import subprocess

print("🌸 [1/4] Memeriksa GPU...")
!nvidia-smi --query-gpu=name,memory.total --format=csv,noheader

print("\n⚡ [2/4] Menginstall Aria2 Fast Downloader...")
!apt-get -y install aria2 > /dev/null 2>&1

print("\n📦 [3/4] Mengunduh KoboldCpp & Model Stheno 8B...")
# Download KoboldCpp binary
if not os.path.exists("koboldcpp-linux-x64"):
    !aria2c -c -x 16 -s 16 "https://github.com/LostRuins/koboldcpp/releases/latest/download/koboldcpp-linux-x64" -o koboldcpp-linux-x64
    !chmod +x koboldcpp-linux-x64

# Model mapping
model_file = f"L3-8B-Stheno-v3.2-{quant_quality}.gguf"
model_url = f"https://huggingface.co/bartowski/L3-8B-Stheno-v3.2-GGUF/resolve/main/{model_file}"

if not os.path.exists(model_file):
    print(f"Mengunduh {model_file} (Kecepatan ~1 Gbps)... Mohon tunggu sebentar.")
    !aria2c -c -x 16 -s 16 "{model_url}" -o "{model_file}"

print(f"\n🔥 [4/4] Menjalankan KoboldCpp (Context: {context_size} tokens, GPU Layers: 99)...\n")
print("👉 Perhatikan URL Cloudflare yang muncul di bawah (format: https://xxxx.trycloudflare.com)!\n")

# Launch KoboldCpp with full GPU offload & remote tunnel
!./koboldcpp-linux-x64 \
  --model "{model_file}" \
  --usecublas normal \
  --gpulayers 99 \
  --contextsize {context_size} \
  --skiplauncher \
  --remotetunnel \
  --host 0.0.0.0 \
  --port 5001
```
